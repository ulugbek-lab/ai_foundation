```python
data_with_missing = {
    "cars": ["BMW", "Volvo", None, "Tesla", "Chevy", None],
    "counts": [None, 7,  None, 6, 2, None],
    "isMarried": [True, True, False, True, False, False]
}
df_missing = pd.DataFrame(data_with_missing)

print(df_missing)
```

```python
df_missing.isnull().sum()
```

```python
df_missing.dropna(how="all")
print(df_missing)
```

```python
df_missing.dropna(subset=['cars'])
```

```python
df_missing.fillna({"cars": "unknown", "counts": 0})
```
