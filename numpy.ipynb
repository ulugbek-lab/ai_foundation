```python
import numpy as np

arr = np.array([1, 2, 3, 5])
print(arr)
```

```python
arr2 = np.array([[1, 2, 3, 5], [4, 6, 8, 9]])
print(arr2)
arr2.shape
```

```python
#dtype()
a = np.array([1,2,3, 7])
print(a.dtype)
```

```python
#reshape()
a = np.array([1,2,3,4,5,6])
b = a.reshape(2,3)
print(b)
```

```python
#astype()
a = np.array([1,2,3])
b = a.astype(float)
print(b)
```

```python
#np.zeros(), np.ones() => dummy data
z = np.zeros((2,3))
o = np.ones((2,3))
print(z)
print(o)
```

```python
#ndim
a = np.array([[1,2], [3,4]])
print(a.ndim)
```

```python
#arrange()
a = np.arange(1, 6)
print(a)
```

```python
#sum(), mean(), max(), min()

a = np.array([1,2,3,4])
print(a.sum())

print(a.mean())

print(a.max())

print(a.min())
```

```python
#flatten()

a = np.array([[1,2], [3,4]])
print(a.flatten())
```

```python
#T(transpose)
a = np.array([[1,2], [3,4]])
print(a.T)
```

```python
#dot() @. (muliplication)
a = np.array([[1,2], [3,4]])
b = np.array([[2,0], [1,2]])

print(a.dot(b))
```

```python
#random.rand()
a = np.random.rand(2,3) #  => 0.0 and 1.0 / 0.0 included / 1.0 excluded
print(a)
```
